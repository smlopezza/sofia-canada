# Connecting to Firestore

```python
import os
from dotenv import load_dotenv
from google.cloud import firestore
```

```python
# Load the .env file
load_dotenv()

# Get the project id from the environment variable
project_id = os.getenv('FIRESTORE_PROJECT_ID')
```

```python
# The `project` parameter is optional and represents which project the client
# will act on behalf of. If not supplied, the client falls back to the default
# project inferred from the environment.
db = firestore.Client(project=project_id)
```

```python
users_ref = db.collection("users")
docs = users_ref.stream()
print("Users in Firestore:", len(list(docs)))
```

```python
# Delete users related to testing

# 1. Convert the string bounds to DocumentReference objects
start_ref = db.collection("users").document("+1555")
end_ref = db.collection("users").document("+1556")

# 2. Target the collection and set up the condition
docs_ref_test = (
    db.collection("users")
    .where(filter=firestore.FieldFilter("__name__", ">=", start_ref))
    .where(filter=firestore.FieldFilter("__name__", "<", end_ref))
)
 
# 3. Stream the matching documents
docs_test = docs_ref_test.stream()
    
# 4. Loop through the documents and delete each one
for doc in docs_test:
    print(f"Deleting document: {doc.id}")
    doc.reference.delete()
```
